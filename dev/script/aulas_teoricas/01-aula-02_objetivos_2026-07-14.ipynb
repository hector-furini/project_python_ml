```python
def func_dados(valor: int | float) -> float:
    """
    Função que recebe um valor e retorna o valor final
        Parametros:
            valor: int | float
        Retorna:
            float
    """
    valor_final = 5 + valor * 2
    return valor_final
```

```python
# Gerando dados aleatórios para teste da função func_dados
n = 200
b = np.random.randint(1, 20, n)
b[:10]
```

```python
# Instânciando nova variável como variação da anterior
a = np.apply_along_axis(func_dados, 0, b)
a[:10]
```

```python
# Instanciando dataframe com os dados gerados
df = pd.DataFrame(data={"x": a, "y": b}).round(0)
df.head()
```

```python
def func_dados_erro(valor: int) -> float:
    """
    Função que recebe um valor e retorna o valor final com erro aleatório
        Parametros:
            valor: int
        Retorna:
            float
    """
    erro = np.random.normal(0, 2, 1)
    valor_final = valor + erro[0]
    return valor_final
```

```python
# Inserindo dados com erro aleatório no dataframe
df["x_noise"] = df["x"].apply(func=func_dados_erro)
df["y_noise"] = df["y"].apply(func=func_dados_erro)
df.head()
```

```python
def novo_valor(valor: int, bias: float, q1y: float, q3y: float) -> float:
    """
    Função que recebe um valor e retorna o valor final com correção
        Parametros:
            valor: int
            bias: float
            q1y: float
            q3y: float
        Retorna:
            float
    """
    if valor <= q1y:
        return valor + bias
    if valor >= q3y:
        return valor + bias
    return valor
```

```python
# Adicionando nova coluna com valores inseridos de bias no dataframe
q1y = df.y_noise.quantile(0.25)
q3y = df.y_noise.quantile(0.75)

df["y_noise_new"] = df["y_noise"].apply(func=novo_valor, args=(5, q1y, q3y))
df.head()
```

```python
# Instanciando variáveis com os valores de min, q1, q3 e max para x e y
minx = df.x_noise.min()
q1x = df.x_noise.quantile(0.25)
q3x = df.x_noise.quantile(0.75)
maxx = df.x_noise.max()

miny = df.y_noise_new.min()
q1y = df.y_noise_new.quantile(0.25)
q3y = df.y_noise_new.quantile(0.75)
maxy = df.y_noise_new.max()

noise = pd.DataFrame(
    {
        "x": [minx, q1x, q3x, maxx],
        "y": [miny, q1y, q3y, maxy],
    }
)
noise
```

```python
def calcular_valor_por_dia(dia: int, valor_inicial: float) -> float:
    """Calcula o valor para um dia específico seguindo a regra:
    - Dias 1 a 4: valor original
    - Dias 5 a 8: incremento de 10% do valor inicial
    - Repete a cada ciclo de 8 dias.
    """
    if dia < 1:
        raise ValueError("O dia deve ser maior ou igual a 1.")

    # O operador % (resto da divisão) nos ajuda a mapear
    # qualquer dia para sua posição equivalente dentro do ciclo de 8 dias (0 a 7)
    posicao_no_ciclo = (dia - 1) % 8

    if posicao_no_ciclo < 4:
        return valor_inicial
    else:
        return valor_inicial * 1.10
```

```python
def __ponto_estrela(df_sample: pd.DataFrame, x: str, y: str) -> list:
    index_superior = df_sample[y][df_sample[y] >= 600].sample(1).index[0]
    retorno = [df_sample.loc[index_superior, x]]
    retorno.append(df_sample.loc[index_superior, y])

    index_inferior = df_sample[y][df_sample[y] < 600].sample(1).index[0]
    retorno.append(df_sample.loc[index_inferior, x])
    retorno.append(df_sample.loc[index_inferior, y])
    return retorno


def seta_estrela(
    df_sample: pd.DataFrame,
    x: str,
    y: str,
    color: str = "g",
    seta: bool = False,
    n_setas: int = 5,
) -> None:
    palette = {"o": "g", "+": "r"}
    sns.scatterplot(
        x=x,
        y=y,
        hue="style_2",
        legend=False,
        s=100,
        palette=palette,
        data=df_sample,
    )
    x_estrela, y_estrela, _, _ = __ponto_estrela(df_sample, x, y)
    plt.plot(x_estrela, y_estrela, "*" + color, markersize=15)
    if seta:
        for _ in range(n_setas):
            index = df_sample[x].sample(1).index[0]
            dx = df_sample[x][index] - x_estrela
            dy = df_sample[y][index] - y_estrela
            plt.arrow(
                x_estrela,
                y_estrela,
                dx,
                dy,
                head_width=1,
                head_length=1,
                length_includes_head=True,
                color=color,
            )
    _, _, x_estrela, y_estrela = __ponto_estrela(df_sample, x, y)
    plt.plot(x_estrela, y_estrela, "*" + color, markersize=15)
    if seta:
        for _ in range(n_setas):
            index = df_sample[x].sample(1).index[0]
            dx = df_sample[x][index] - x_estrela
            dy = df_sample[y][index] - y_estrela
            plt.arrow(
                x_estrela,
                y_estrela,
                dx,
                dy,
                head_width=1,
                head_length=1,
                length_includes_head=True,
                color=color,
            )
    return None
```

```python
grafico = {
    "df_sample": dados,
    "x": "x_noise",
    "y": "y_noise_new_cal",
    "seta": True,
}
seta_estrela(**grafico)
grafico = {
    "df_sample": dados,
    "x": "x_noise",
    "y": "y_noise_new_cal",
    "color": "r",
    "seta": True,
}
obsr = "Gramas"
pred = "Calorias"
seta_estrela(**grafico)
plt.title("Scatter Plot de X/Y")
plt.xlabel(obsr)
plt.ylabel(pred)
plt.show()
```
